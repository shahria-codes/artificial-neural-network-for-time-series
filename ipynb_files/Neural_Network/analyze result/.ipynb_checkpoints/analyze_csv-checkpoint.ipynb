```python
import pandas as pd
import os

# ==================================================
# CONFIGURATION – ADJUST COLUMN NAMES IF NEEDED
# ==================================================
TEST_COL = "test %"          # column for test percentage
VAL_COL   = "val %"          # column for validation percentage
SEED_COL  = "seed"
H_SIZE_COL = "h_size"
LRATE_COL = "l_rate"

# Path to the folder containing results_0.csv ... results_20.csv
DATA_FOLDER = "../seed_result"

# Output file name
OUTPUT_CSV = "filtered_results.csv"

# ==================================================
# SCRIPT STARTS HERE
# ==================================================

# Generate the list of file paths from results_0.csv to results_20.csv
csv_files = [os.path.join(DATA_FOLDER, f"results_{i}.csv") for i in range(21)]

# Verify which files actually exist
existing_files = [f for f in csv_files if os.path.exists(f)]
missing_files = [f for f in csv_files if not os.path.exists(f)]

print(f"Expected 21 files, found {len(existing_files)}.")
if missing_files:
    print(f"Missing: {missing_files}")

if not existing_files:
    print("No files found. Check folder name and file names.")
    exit()

# List to hold dataframes
dfs = []

for file in existing_files:
    try:
        df = pd.read_csv(file)
        dfs.append(df)
    except Exception as e:
        print(f"Error reading {file}: {e}")

# Combine all data
if not dfs:
    print("No valid dataframes to combine.")
    exit()

combined = pd.concat(dfs, ignore_index=True)
print(f"Total rows combined: {len(combined)}")

# Check for required columns
required_cols = [TEST_COL, VAL_COL, SEED_COL, H_SIZE_COL, LRATE_COL]
missing_cols = [col for col in required_cols if col not in combined.columns]
if missing_cols:
    print(f"ERROR: Missing columns in CSV files: {missing_cols}")
    print(f"Available columns: {list(combined.columns)}")
    exit()

# Convert test % to numeric (handles strings or missing values)
combined[TEST_COL] = pd.to_numeric(combined[TEST_COL], errors='coerce')

# Filter rows where test % >= 70
filtered = combined[combined[TEST_COL] >= 70].copy()
print(f"Rows with {TEST_COL} >= 70: {len(filtered)}")

# Select desired columns with test % as the leading column
result = filtered[[TEST_COL, SEED_COL, H_SIZE_COL, LRATE_COL, VAL_COL]]

# Display results in console
print("\n--- Filtered results (test %, seed, h_size, l_rate, val %) ---")
print(result.to_string(index=False))

# Save to CSV
result.to_csv(OUTPUT_CSV, index=False)
print(f"\nResults saved to {OUTPUT_CSV}")
```

```python
import pandas as pd
import os

# ==================================================
# CONFIGURATION – ADJUST COLUMN NAMES IF NEEDED
# ==================================================
TEST_COL = "test %"          # column for test percentage
VAL_COL   = "val %"          # column for validation percentage
SEED_COL  = "seed"
H_SIZE_COL = "h_size"
LRATE_COL = "l_rate"

# Path to the folder containing results_0.csv ... results_20.csv
DATA_FOLDER = "../seed_result"

# Output file name
OUTPUT_CSV = "filtered_results_val_test.csv"

# ==================================================
# SCRIPT STARTS HERE
# ==================================================

# Generate file paths from results_0.csv to results_20.csv
csv_files = [os.path.join(DATA_FOLDER, f"results_{i}.csv") for i in range(21)]

# Verify existing files
existing_files = [f for f in csv_files if os.path.exists(f)]
missing_files = [f for f in csv_files if not os.path.exists(f)]

print(f"Expected 21 files, found {len(existing_files)}.")
if missing_files:
    print(f"Missing: {missing_files}")

if not existing_files:
    print("No files found. Check folder name and file names.")
    exit()

# Read all CSVs
dfs = []
for file in existing_files:
    try:
        df = pd.read_csv(file)
        dfs.append(df)
    except Exception as e:
        print(f"Error reading {file}: {e}")

if not dfs:
    print("No valid dataframes to combine.")
    exit()

combined = pd.concat(dfs, ignore_index=True)
print(f"Total rows combined: {len(combined)}")

# Check required columns
required_cols = [TEST_COL, VAL_COL, SEED_COL, H_SIZE_COL, LRATE_COL]
missing_cols = [col for col in required_cols if col not in combined.columns]
if missing_cols:
    print(f"ERROR: Missing columns in CSV files: {missing_cols}")
    print(f"Available columns: {list(combined.columns)}")
    exit()

# Convert both percentage columns to numeric
combined[TEST_COL] = pd.to_numeric(combined[TEST_COL], errors='coerce')
combined[VAL_COL] = pd.to_numeric(combined[VAL_COL], errors='coerce')

# Apply both conditions: test% >= 70 AND val% > 50
filtered = combined[(combined[TEST_COL] >= 70) & (combined[VAL_COL] > 50)].copy()
print(f"Rows with {TEST_COL} >= 70 and {VAL_COL} > 50: {len(filtered)}")

# Select columns with test % as leading column
result = filtered[[TEST_COL, SEED_COL, H_SIZE_COL, LRATE_COL, VAL_COL]]

# Display results in console
print("\n--- Filtered results (test %, seed, h_size, l_rate, val %) ---")
print(result.to_string(index=False))

# Save to CSV
result.to_csv(OUTPUT_CSV, index=False)
print(f"\nResults saved to {OUTPUT_CSV}")
```
