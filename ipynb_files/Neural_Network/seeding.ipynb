```python
import papermill as pm
import tempfile
import os

# Temporary folder auto-deleted after the 'with' block
with tempfile.TemporaryDirectory() as tmpdir:
    for seed in range(100, 201):
        out_path = os.path.join(tmpdir, f"temp_seed_{seed}.ipynb")
        pm.execute_notebook(
            input_path="ffnn_different_seed.ipynb",
            output_path=out_path,                  # safely inside tmpdir
            parameters={"seed_value": seed},
            kernel_name="python3"
        )
        print(f"Finished seed={seed}")
# All temp files are now gone
```

```python
import os
import tempfile
import papermill as pm

# Define the ranges or lists for your parameters
seeds = range(1, 101)
neuron_sizes = range(4, 20)          # Example sizes
learning_rates = [0.01, 0.05, 0.08, 0.001, 0.002 0.0001] # Example learning rates

with tempfile.TemporaryDirectory() as tmpdir:
    for seed in seeds:
        for neuron_size in neuron_sizes:
            for lr in learning_rates:
                
                # Create a unique filename for this specific combination
                filename = f"temp_s{seed}_n{neuron_size}_lr{lr}.ipynb"
                out_path = os.path.join(tmpdir, filename)
                
                pm.execute_notebook(
                    input_path="ffnn_different_seed.ipynb", 
                    output_path=out_path,
                    parameters={
                        "seed_value": seed,
                        "neuron_size": neuron_size,     # Pass neuron_size
                        "learning_rate": lr             # Pass learning_rate
                    },
                    kernel_name="python3"
                )
                print(f"Finished seed={seed} | neurons={neuron_size} | lr={lr}")

# All temp files are now gone
```
